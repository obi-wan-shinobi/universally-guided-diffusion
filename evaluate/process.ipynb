```python
import os
from PIL import Image

def create_image_grids(root_dir: str, output_name: str = "grid.png"):
    """
    For each subdirectory under root_dir:
    - Read six images
    - Arrange them in a 2x3 grid (2 rows, 3 columns)
    - Save the resulting image as output_name in the same subdirectory
    Returns a list of saved grid image paths.
    """
    saved_paths = []
    for entry in os.scandir(root_dir):
        if not entry.is_dir():
            continue
        sub_path = entry.path
        # collect image files
        img_files = [os.path.join(sub_path, f) for f in os.listdir(sub_path)
                     if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp', '.gif', '.tiff'))]
        img_files.sort()
        if len(img_files) != 6:
            print(f"Warning: {sub_path} contains {len(img_files)} images, skipping")
            continue
        # open images
        imgs = [Image.open(p) for p in img_files]
        # normalize size to first image
        w, h = imgs[0].size
        imgs = [im.resize((w, h), resample=Image.Resampling.LANCZOS) for im in imgs]
        # create empty canvas
        grid_img = Image.new('RGB', (w * 3, h * 2))
        # paste images in row-major order
        for idx, im in enumerate(imgs):
            row, col = divmod(idx, 3)
            grid_img.paste(im, (col * w, row * h))
        save_path = os.path.join(sub_path, output_name)
        grid_img.save(save_path)
        saved_paths.append(save_path)
    return saved_paths
```

```python
create_image_grids("eval_data/gt-ddim-50/")
```
